```python
df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0, 3000, 6000, 10000, 15000, 25000],
    labels=["0-3000", "3000-6000", "6000-10000",
            "10000-15000", "15000-25000"]
)

pd.crosstab(
    df["IncomeGroup"],
    df["Attrition"],
    normalize="index"
).mul(100).round(2)
```
